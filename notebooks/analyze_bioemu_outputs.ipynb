# Analyze Precomputed BioEmu Outputs

Use this notebook after running BioEmu in Colab and downloading or syncing the portable bundle locally.

```python
from pathlib import Path

from guided_conformational_path.bioemu_experiment import (
    BioEmuPrecomputedConfig,
    run_bioemu_precomputed_experiment,
)
from guided_conformational_path.large_synthetic import (
    display_py3dmol_path,
    format_run_summary,
    write_run_outputs,
)
from IPython.display import Image, display
```

## Configure Paths

Place endpoint PDBs at `data/pdb_inputs/A.pdb` and `data/pdb_inputs/B.pdb`. Place the BioEmu bundle at `data/bioemu_outputs/latest` or update `bioemu_bundle_dir` below.

```python
config = BioEmuPrecomputedConfig(
    endpoint_a_path=Path("data/pdb_inputs/A.pdb"),
    endpoint_b_path=Path("data/pdb_inputs/B.pdb"),
    bioemu_bundle_dir=Path("data/bioemu_outputs/latest"),
    max_bioemu_samples=200,
    outputs_dir=Path("outputs/bioemu_experiment"),
    write_nodes=False,
    write_pymol=True,
    write_py3dmol=True,
    write_animation=True,
)
config
```

## Run Graph Analysis

BioEmu samples are treated as candidate graph nodes only. The shortest path is a heuristic structural path, not a validated physical transition pathway.

```python
run = run_bioemu_precomputed_experiment(config)
write_run_outputs(run, config)
print(format_run_summary(run, config.outputs_dir))
```

## Inspect Outputs

```python
gif_path = run.summary.get("path_morph_gif")
if gif_path:
    display(Image(filename=gif_path))

display_py3dmol_path(run)
```
