# Analyze A Downloaded Dataset BioEmu Bundle

Use this concise local UI after downloading a Colab bundle. It uses the exported dataset pair manifest and residue mapping rather than requiring manually prepared `A.pdb` and `B.pdb` files.

```python
from pathlib import Path
from IPython.display import Image, display

from guided_conformational_path.dataset_bioemu_experiment import (
    DatasetBioEmuConfig,
    run_dataset_bioemu_experiment,
)
from guided_conformational_path.large_synthetic import format_run_summary, write_run_outputs
```

## Configure Paths

Download the Colab bundle into `data/bioemu_outputs/first_five_proteins/<protein>/<pair>/n_<count>`, or change the path below.

```python
PROTEIN_ID = "P69441"
PAIR_ID = "4AKE_A__1AKE_A"
REQUESTED_SAMPLES = 200

config = DatasetBioEmuConfig(
    dataset_pair_dir=Path("data/dataset/first_five_proteins") / PROTEIN_ID / "pairs" / PAIR_ID,
    bioemu_bundle_dir=(
        Path("data/bioemu_outputs/first_five_proteins")
        / PROTEIN_ID
        / PAIR_ID
        / f"n_{REQUESTED_SAMPLES}"
    ),
    max_bioemu_samples=REQUESTED_SAMPLES,
    k_neighbors=10,
    outputs_dir=Path("outputs/dataset_bioemu") / PROTEIN_ID / PAIR_ID / f"n_{REQUESTED_SAMPLES}",
    write_pymol=True,
    write_py3dmol=True,
    write_animation=True,
)
print(config)
```

## Run Graph Analysis

Sequence and residue-count mismatches fail before graph construction. BioEmu nodes are candidate structures only; the resulting path is not a kinetic pathway.

```python
run = run_dataset_bioemu_experiment(config)
write_run_outputs(run, config)
print(format_run_summary(run, config.outputs_dir))
```
